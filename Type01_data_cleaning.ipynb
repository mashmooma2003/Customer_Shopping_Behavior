```python
import pandas as pd

# Load the dataset
df = pd.read_csv('customer_shopping_behavior.csv')

# Clean column names (lowercase with underscores)
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Display the first 5 rows
df.head()
```

```python
# Check data types, non-null counts, and missing values
print("--- DATASET INFO ---")
df.info()

print("\n--- MISSING VALUES ---")
print(df.isnull().sum())

print("\n--- DUPLICATE ROWS ---")
print(f"Total duplicate rows: {df.duplicated().sum()}")
```

```python
# 1. Fill missing numerical values with median (if any exist)
# 2. Drop duplicates
df = df.drop_duplicates()

# 3. Rename columns to standard SQL-friendly snake_case format
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')

# Verify the cleaned structure
df.head()
```

```python
# Save cleaned data to CSV
df.to_csv('cleaned_customer_shopping_behavior.csv', index=False)
print("Cleaned dataset successfully saved as 'cleaned_customer_shopping_behavior.csv'!")
```

```python
import duckdb

# Query 1: Total Revenue and Average Purchase Amount by Gender
query1 = """
SELECT gender, 
       COUNT(*) AS total_customers, 
       ROUND(SUM(purchase_amount_usd), 2) AS total_revenue,
       ROUND(AVG(purchase_amount_usd), 2) AS avg_purchase
FROM df
GROUP BY gender;
"""

# Query 2: Top 5 Most Popular Item Categories by Purchase Count
query2 = """
SELECT category, 
       COUNT(*) AS total_purchases, 
       ROUND(SUM(purchase_amount_usd), 2) AS total_sales
FROM df
GROUP BY category
ORDER BY total_purchases DESC
LIMIT 5;
"""

print("--- Revenue by Gender ---")
display(duckdb.sql(query1).df())

print("\n--- Top 5 Categories ---")
display(duckdb.sql(query2).df())
```
